```python
#What to do about missing data -> 
#children impute missing value with zero(since they are so few missing values and most of the data shows that customers are not children 
#and/or with children)
impute_with_zero = (hotel['children'].isna() == True)
impute_with_zero = impute_with_zero[impute_with_zero == True]
hotel.loc[impute_with_zero.index, 'children'] = 0
hotel['children'].isna().sum()
```

```python
#How to deal with country 
hotel['country'] = hotel['country'].fillna('Unknown') #fill it with unknown as it shouldn't impact the data too much
hotel['country'].isna().sum()
```

```python
#How to deal with agent feature
hotel['agent'] = hotel['agent'].fillna(-1) #negative one means that the 
hotel['agent'].isna().sum()
```

```python
#25. company -> columns contains ID of the company that made the booking  
#feature 25 has over 90% missing data so we can create a binary variable that shows whether the account is a company account

company_binary = (hotel['company'].isna() == True)
company_binary = company_binary[company_binary == True]
company_binary
hotel.loc[company_binary.index, 'company'] = 0
#hotel['company'].value_counts()
company_false = hotel['company'] != 0.0
#company_false.sum()
hotel.loc[company_false, 'company'] = 1
hotel['company'] = hotel['company'].astype('int')
hotel['company']
#hotel['company'].value_counts()
```

```python
hotel.isna().sum() #no more missing data!!!
```

### Dropping columns

```python
#dropping these two columns as the information they provid
#does not seem significant enough considering we have
#arrival_date_year and arrival_date_month
hotel = hotel.drop(columns=['arrival_date_week_number'
                    ,'arrival_date_day_of_month'])
hotel.head()
```

```python
#dropping reservation_status and reservation_status_date because both of these columns allude to directly
#predicting the target variable is_canceled which we do not want a model with strong features that tell us
#the outcome based on these two features alone (not helpful for the business to improve hotel cancellation rate)
hotel = hotel.drop(columns=['reservation_status'
                    ,'reservation_status_date'])
hotel.head()
```

```python
#for features stays_in_weekend_nights and stays_in_week_nights we can feature engineer these two columns into 
#one where it is a binary feature on whether they stay the weekend or not
hotel['stays_in_weekend_nights'].value_counts()
weekend_nights_yes = hotel['stays_in_weekend_nights'] > 0
weekend_nights_yes = weekend_nights_yes[weekend_nights_yes == True]
hotel.loc[weekend_nights_yes.index, 'stays_in_weekend_nights'] = 1
hotel['stays_in_weekend_nights'].value_counts()

#drop stay_in_week_nights column as it is no longer needed
hotel = hotel.drop(columns=['stays_in_week_nights'])
hotel.head()
```

```python
#for features children and babies consider combining them into one group since individual they act the same
#(running this code multiple times will change results
hotel['children'] = hotel['children'] + hotel['babies']
hotel['children'] = hotel['children'].astype('int')

hotel = hotel.drop(columns=['babies'])
hotel.head()
```

```python
#reserved_room feature worth discussing today
```

```python
#adr 0 = budget, 1 = standard, 2 = premium 
hotel['adr'].value_counts()

adr_budget = hotel['adr'] <= 60.00
adr_budget = adr_budget[adr_budget == True]
adr_budget[adr_budget == True] = 0.0
adr_budget

adr_standard = (hotel['adr'] > 60.00) & (hotel['adr'] <= 120.00)
adr_standard = adr_standard[adr_standard == True]
adr_standard[adr_standard == True] = 1.0
adr_standard

adr_premium = hotel['adr'] > 120.00
adr_premium = adr_premium[adr_premium == True]
adr_premium[adr_premium == True] = 2.0
adr_premium

hotel.loc[adr_budget.index, 'adr'] = 0.0
hotel.loc[adr_standard.index, 'adr'] = 1.0
hotel.loc[adr_premium.index, 'adr'] = 2.0
hotel['adr'] = hotel['adr'].astype('int')
#hotel['adr'].value_counts()
```
